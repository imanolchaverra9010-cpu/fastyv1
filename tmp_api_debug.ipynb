```python
# Debug de Fetch /api/businesses 500
Este notebook ayuda a reproducir y analizar el error de `GET /api/businesses?status_filter=active` en producción.
</VSCode.Cell>
<VSCode.Cell language="python">
# 1. Set up the notebook environment
import requests
import json
from urllib.parse import urlencode

print('Environment ready')
</VSCode.Cell>
<VSCode.Cell language="python">
# 2. Reproduce the API fetch request
base_url = 'https://fasty-delta.vercel.app/api/businesses'
params = {'status_filter': 'active'}
full_url = f"{base_url}?{urlencode(params)}"
print('Request URL:', full_url)

response = requests.get(full_url, timeout=15)
print('Response status:', response.status_code)
print('Response headers:', dict(response.headers))
</VSCode.Cell>
<VSCode.Cell language="python">
# 3. Inspect the HTTP response and console error
try:
    payload = response.text
except Exception as exc:
    payload = f'<unable to read body: {exc}>'

print('Response body:')
print(payload[:2000])
</VSCode.Cell>
<VSCode.Cell language="python">
# 4. Handle fetch errors and status codes
if response.ok:
    try:
        data = response.json()
        print('Parsed JSON response:')
        print(json.dumps(data, indent=2)[:2000])
    except Exception as exc:
        print('JSON parsing failed:', exc)
else:
    print('Non-200 response detected')
    print('Status code:', response.status_code)
    print('Response text:', response.text)
</VSCode.Cell>
<VSCode.Cell language="python">
# 5. Log structured request/response details
print('Request details:')
print('URL:', response.request.url)
print('Method:', response.request.method)
print('Request headers:', dict(response.request.headers))
print('Response status:', response.status_code)
print('Response reason:', response.reason)
</VSCode.Cell>
<VSCode.Cell language="python">
# 6. Simulate a server error and debug with mock data
from requests.models import Response

mock = Response()
mock.status_code = 500
mock._content = b'A server error has occurred'

print('Mocked response ok:', mock.ok)
print('Mocked response text:', mock.text)
try:
    data = mock.json()
except Exception as exc:
    print('Mock parse error:', exc)
```
